### Residual Analysis

Looking at the residual vs Predicted graphs for the models, one thing becomes quite clear. The models' predictions seem to barely consider the largest fires. Most of the residual error stems from a couple of of very large differences. This reveals why the median baseline model remains the most reliable, all of the models are already very wrong when predicting large fires, as such, the thing that differentiates the performance of the models is reliably predicting that very little fire will occur. With the median being 0.54, the median model simply performs better than the others because it consistently guesses a value that's very close to 0.
