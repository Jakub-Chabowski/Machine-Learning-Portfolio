### Milestone 3 conclusion
The Baseline Model, Median, remains the most reliable one, having the best scores across MAE, RMSE, R2 and residual scores.
However, the Nerual Network Models using MLPRegressor seem to be getting similar results, with them sometimes getting slightly better results than the baseline in certain categories (mostly in the r2 score). However, this could be attributed to noise due to the exteremly small values that I'm getting for R2.
Perhaps with some more tuning they can surpass the baseline. So far, tuning these models has included changing the hidden layer sizes, changing the solver and changing the random state, with no meaningful changes in the final results.
For the other models, results haven't been notable either. 
Increasing the max depth in the random forests doesn't seem to make them perform any better, only increasing the time taken.

All of these problems seem to come back to the extreme skew in the data. From looking at the scatter plots between the predicted and actual values it's clear that the models don't ever guess the higher values for the area burned, which leads to massive residuals. This is probably due to the small amount of actual large values, making them act more like outliers despite their importance. I suspect that this could possibly be due to a lack of examples of large fires in either the training or test sets, perhaps I can try using some sort of StratifiedKFold to make sure that both sets retain similar ratios of large fires, small fires and no fires. This is something I tried before but couldn't get to work reliably, it might be useful to revisit this.
